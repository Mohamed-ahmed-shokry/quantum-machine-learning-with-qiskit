# 1. Quantum circuits and entanglement

This first lab builds a Bell state and samples it with Qiskit's V2 `StatevectorSampler`. It replaces the legacy `Aer.get_backend(...)` and `execute(...)` workflow.

## Build the circuit

A Hadamard gate creates a superposition on the first qubit. A controlled-X gate then entangles both qubits, so measurements should be split between `00` and `11`.

```python
from qiskit import QuantumCircuit
from qiskit.primitives import StatevectorSampler
```

```python
circuit = QuantumCircuit(2)
circuit.h(0)
circuit.cx(0, 1)
circuit.measure_all()

print(circuit.draw(output="text"))
```

## Sample the Bell state

The fixed seed makes this finite-shot experiment reproducible. Small differences between the two counts are expected sampling noise.

```python
sampler = StatevectorSampler(seed=42)
result = sampler.run([(circuit,)], shots=1024).result()[0]
counts = result.data.meas.get_counts()
counts
```

```python
assert set(counts) == {"00", "11"}
assert sum(counts.values()) == 1024
```

## Try it yourself

Remove the controlled-X gate and predict how the counts change. Then increase the shot count and observe how the measured proportions approach their theoretical values.
